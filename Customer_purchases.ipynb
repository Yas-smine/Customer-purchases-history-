    The objective is to find the association rules to help supermarket owners find new marketing plans to improve their sales.

```python
toy_dataset = [['Skirt', 'Sneakers', 'Scarf', 'Pants', 'Hat'],

    ['Sunglasses', 'Skirt', 'Sneakers', 'Pants', 'Hat'],

    ['Dress', 'Sandals', 'Scarf', 'Pants', 'Heels'],

    ['Dress', 'Necklace', 'Earrings', 'Scarf', 'Hat', 'Heels', 'Hat'],

    ['Earrings', 'Skirt', 'Skirt', 'Scarf', 'Shirt', 'Pants']]
```

Run the apriori algorithm on the provided toy_dataset. Interpret the results.

```python
import mlxtend
import pandas as pd 
from mlxtend.preprocessing import TransactionEncoder  # Using `TransactionEncoder`
TE = TransactionEncoder()
TE_ary= TE.fit_transform(toy_dataset)  #Apply one-hot-encoding on our toy_dataset
TE_ary
```

```python
toy_df= pd.DataFrame(TE_ary, columns= TE.columns_) #Creating a new DataFrame from our Numpy array
toy_df
```

Now that the dataset has been transformed using one-hot encoding, we can move on to **extracting frequent itemsets** using the **Apriori algorithm**.

```python
from mlxtend.frequent_patterns import apriori
apriori(toy_df, min_support=0.6) # Support code 
```

```python
frequent_itemsets=apriori(toy_df, min_support=0.02, use_colnames=True)  #Instead of column indices we can use column names
frequent_itemsets #Support with column names
```

Once we have our frequent itemsets, we can extract **association rules** using other interesting metrics like **confidence** and **lift**.

```python
from mlxtend.frequent_patterns import association_rules
association_rules(frequent_itemsets, metric="confidence", min_threshold=0.01) # associate items
```

In addition to **confidence**, we can also use **lift** as a metric to extract more meaningful association rules.

```python
from mlxtend.frequent_patterns import association_rules
association_rules(frequent_itemsets,metric="lift",min_threshold=1.25)
```

    Top Association Rules Interpretation
**1. (Skirt) → (Pants)**

- Support: 0.6, This pair appears in 60% of all transactions.

- Confidence: 1.0, Every time Skirt appears, Pants also appears.

- Lift: 1.25, lift >1, so the likelihood of Pants being bought increases by 25% when Skirt is present.

- Certainty: 1.0, Strongest possible certainty that Pants follows Skirt.

This is a very strong rule. If a customer buys a Skirt, they always buy Pants too. Great for bundling or recommendation systems.

**2. (Pants) → (Skirt)**

- Support: 0.6

- Confidence: 0.75, 75% of Pants purchases also include Skirt.

- Lift: 1.25, Still a positive association.

- Certainty: 0.375, Lower than the reverse rule.

While Pants often lead to Skirt, it's not guaranteed. Still a good cross-sell opportunity, but not as strong as the reverse.

**3. (Scarf) → (Pants) and (Pants) → (Scarf)**

- Support: 0.6

- Confidence: 0.75 in both directions

- Lift: 0.9375, Slightly below 1, which suggests a negative or neutral association

- Certainty: -0.25 , Weak and potentially misleading rule.

Despite high confidence, the lift < 1 and negative leverage suggest this rule is not truly useful. These items co-occur often, but not more than random chance would predict.

```python
import csv # Imports the csv module to handle CSV file operations

# Defines a function to read a CSV file and return its contents as a list of lists
def read_csv_as_list_of_lists(filename):
    data = []  # Initializes an empty list to store rows from the CSV
    with open(filename, newline='', encoding='utf-8') as csvfile: # Opens the CSV file in read mode
        reader = csv.reader(csvfile)  # Creates a CSV reader object
        for row in reader:
            # Checks if the row is not empty before appending
            if row:
                data.append(row) # Adds the non-empty row to the data list
    return data  # Returns the complete list of rows

# Example usage
filename = 'C:\\Users\\ABL\\Documents\\Codes\\checkp_python\\ML\\Association_rules_checkp\\Customer_purchases_history\\DATASET\\Market_Basket_Optimisation.csv'  # Replace with your actual file path
rows = read_csv_as_list_of_lists(filename)  # Calls the function and stores the result in 'rows'
rows # Displays the loaded data (each row is a list of items purchased by a customer)
```

```python
# Flatten list of lists into one pandas Series
df = pd.Series([item for sublist in rows for item in sublist], name="Item")

# Count frequencies
freq = df.value_counts().reset_index()
freq.columns = ['Items', 'Count']
```

```python
# Initialize an empty list to store item pairs
pairs = []

# Loop through each transaction (basket) in the dataset
for transaction in rows:
    # Remove duplicate items within the same basket
    unique_items = list(set(transaction))
    
    # Generate all unique pairs of items from the basket
    for i in range(len(unique_items)):
        for j in range(i + 1, len(unique_items)):
            # Sort the pair alphabetically to avoid duplicates like ['A', 'B'] and ['B', 'A']
            pair = sorted([unique_items[i], unique_items[j]])
            # Add the pair to the list
            pairs.append(pair)

# Convert the list of pairs into a DataFrame with two columns: Item1 and Item2
df_pairs = pd.DataFrame(pairs, columns=['Item1', 'Item2'])

# Count how many times each unique pair appears in the dataset
pair_counts = df_pairs.value_counts().reset_index(name='Count')

# Select the top 10 most frequent item pairs
top_pairs = pair_counts.head(10)

# Create a horizontal bar chart using Plotly
fig = px.bar(
    top_pairs,  # DataFrame containing top item pairs and their counts
    x='Count',  # Frequency of each pair on the x-axis
    y=top_pairs['Item1'] + ' + ' + top_pairs['Item2'],  # Concatenate item names for y-axis labels
    orientation='h',  # Horizontal bars
    text='Count',  # Show count values on each bar
    title='Top 10 Most Frequently Bought-Together Pairs (No itertools)',  # Chart title
    color='Count',  # Color bars based on frequency
    color_continuous_scale='Blues'  # Use a blue gradient for visual appeal
)

# Position the count labels outside the bars
fig.update_traces(textposition='outside')

# Customize axis titles
fig.update_layout(yaxis_title='Item Pair', xaxis_title='Frequency')

# Display the chart
fig.show()
```
